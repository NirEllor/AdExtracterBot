```python
# Creates and runs a single report depending based on brand name
def create_and_run_rep(brand_name: str):
    # Open file with relevant parameters to create report
    with open('creatives_payload.json') as user_file:
        file_contents = user_file.read()

    payload = json.loads(file_contents)

    # Update parameters to match brand
    payload['Title'] = brand_name
    search_payload['SearchTerm'] = brand_name

    # Create and run report
    brand_search = requests.post('https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/Entity/EntitySearch', headers=headers, json=search_payload, verify=False).content
    brand_res = json.loads(brand_search.decode('utf-8'))['Results']

    # Find the correct search entity of the brand
    res = None
    for key in brand_res:
        if key['EntityName'].lower() == brand_name.lower():
            res = key

    if res is None:
        return

    # Update the payload to match the search entity of brand
    payload['BrandCentralSelections']['Entities'] = [res]
    payload['ExcludedSearchEntityItems'] = [res]

    # Create and run report
    create_report = requests.post('https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/CreateReportSpec', headers=headers, verify=False)
    update_report = requests.post('https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/UpdateReportSpec', headers=headers, json=payload, verify=False)
    save_report = requests.post('https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/SaveReportSpec', headers=headers, data=brand_name+'*', verify=False)
    report_spec_id = save_report.text
    run_report = requests.post(f'https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/DoReportSpecAction?reportSpecId={report_spec_id}&action=Run', headers=headers, data=f'reportSpecId={report_spec_id}&action=Run', verify=False)

    return report_spec_id

# Creates multiple reports given the starting and ending index to take from the brands list
def create_multiple_reps(start, end):
    print('Creating reports')
    for brand in brands[start:end]:
        try:
            create_and_run_rep(brand)
        except Exception as e:
            print(e)

    # Wait until reports are completed
    print('Reports created, waiting for reports to run')
    while not check_reports_status():
        time.sleep(60)
    

# Checks if all reports have completed running
def check_reports_status():
    # Get current reports
    get_reports = requests.get(
        'https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/GetReportListData?mode=FullView',
        headers=headers, data='FullView', verify=False)
    time.sleep(3)
    reps = json.loads(get_reports.text)

    # Check status
    for key in reps['ReportList']['Reports']:
        if key['Status'] != 2: # Uncompleted report
            return False

    return True

# Gets download link of reports created
def get_download_links(brand_names):
    links = []

    # Get all reports
    get_reports = requests.get('https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/api/CustomReporting/GetReportListData?mode=FullView', headers=headers, data='FullView', verify=False)
    time.sleep(3)
    reps = json.loads(get_reports.text)

    # Select relevant reports
    for key in reps['ReportList']['Reports']:
        if key['Status'] == 2 and any(map(key['ReportName'].__contains__, brand_names)):
            report_id = key['ReportId']
            name = key['ReportName'].rsplit(' ', 1)[0]
            creative_link = f'https://app.vivvix.com/360/KMI/IntelliDrive/ApplUI/Pages/CreativePages/DownloadCreatives.aspx?.pl=DownloadCreatives&searchType=CustomReporting&reportId={report_id}'
            csv_link = key['ReportOutputs'][0]['FullFilePath']
            links.append({'Creative': creative_link, 'CSV': csv_link, 'reportID': report_id, 'Brand': name})

    return links


# Downloads all reports through their links
def download_reports(start, end, path):
    # Get download links
    links = get_download_links(brands[start:end])
    print('Downloading creatives')
    
    # Creates folder to download creatived incase it doesn't exist
    if not os.path.exists(path):
        os.makedirs(path) 

    # Create request header for creatives download
    download_header = headers
    download_header['Accept'] = 'text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8,application/signed-exchange;v=b3;q=0.7'
    download_header['Accept-Encoding'] = 'gzip, deflate, br'
    
    # Download creative brand by brand
    for item in links:
        # Create relevant folder to brand
        # path = os.path.join("C:/Users/Admin/Documents/Lab/Vivvix Script/creatives/", item['Brand'].replace('*', '')).strip()
        download_path = os.path.join(path, item['Brand'].replace('*', '')).strip()    
        os.mkdir(download_path)
        
        # Download creatives
        # zip_req = requests.post(item['Creative'], headers=download_header, verify=False)
        # open(f'{download_path}/Creatives.zip', 'wb').write(zip_req.content)
        
        # Download report
        wget.download(item['CSV'], out=download_path)
    
    print('Downloading done')
 

# Runs the whole process at once
def run_batch(start, end, download_path):
    # Creates report for every brand
    create_multiple_reps(start, end)

    # Download creatives 
    # links = get_download_links(brands[start:end])
    download_reports(start, end, download_path)
```
